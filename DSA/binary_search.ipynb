Defining the binary search algo as a stand alone func

```python
# Binary search RELIES ON A SORTED LIST or might retun none 
# even if the value is within the list - has to be sorted in asc order
# It breaks the array / list into subarrays / sublists 
# it compares the value at the middle of that sublist 
# until the target value is found 
# Therefore we need a way to keep track of the position 
# of the list we're working with - first and lat to point to 
# the beginning and end of the array 
# an iterative solution - aka uses a loop 
# not a functional language as creating first and last vars and then 
# modifying it until reaching the solution is not functional as functional 
# language does not like modifying the value of vars within the function 

def binary_search(list_being_searched_through, target_value):
    """
    Returns the index position if the target is found 
    Otherwise returns none 
    """
    # instantiate first and last objs to keep track of 
    # the psotion we're at as we go through each sublist 

    first = 0 # O(n)
    last = len(list_being_searched_through) - 1 # O(n) eventhoguh using len func 

    # Using a while loop as opposed to the for loop to
    # ensure the algo keeps running 
    # In bs, we use a while loop until first is lesss than 
    # or equal to the value of last 
    # In an empty lsit, first is greater than last thf loop would not execute 
    while first <= last: # O(log n)
        # Calc the mid point of the list as that's the first 
        # step in bs using floor division to get whole num - 
        # rounds down e.g. 3.5 -> 3 
        midpoint = (first + last) // 2 # O(n)

        # Check if the midpoint valuewithin the list = target_val
        # Best case scenario
        if list_being_searched_through[midpoint] == target_value: # O(n)
            return midpoint 
        elif list_being_searched_through[midpoint] < target_value: # O(n)
            # If the midpoint value, is smaller than the actaul target value, 
            # we don't care about any vals greater than it and then reposition 
            # the start of outr list / first point to be the midpoint of the new 
            # sublist + 1 as we know it's definitely not the midpoint 
            # not actually creating a sublist but redef first and last 
            # as the algo only considers the vals b/w first and last when determine
            # the midpoint  when redef first and last as the algo proceeds, we can 
            # find the midpoint using the original list resulting in constant runtime
            # in space complexity 
            first = midpoint + 1
        else: # O(n)
            # If the val at the midpoint is greater than the target, discard 
            # the vals after the midpoint and redef last to be the midpoint -1
            last = midpoint - 1

    # Never hit a case where the midpoint = target as the list did not contain
    # the target thf return none
    return None

# Logarithmic  run time as we ask the algo to eval as many times as it needs 
# until first is equal/greater than last 
# Each time the loop does this, the size of the sublist grows smaller until it
# reaches a single element This is what creates the ln rt
```

```python
# Cleaned up 

# What makes it an algo: must return a value, 
# must complete execution in a finite amount of time, 
# must return the same output for the same given input

def binary_search(list, target):
    """
    Returns the index position of the target if found 
    Ohterwise returns none 
    """

    first = 0 
    last = len(list) - 1 

    while first <= last:
        midpoint = (first + last) // 2

        if list[midpoint] == target:
            return midpoint 
        elif list[midpoint] < target:
            first = midpoint + 1
        else:
            last = midpoint - 1
    
    return None
```
