##### Introduction to Algorithm - Playing a Counting Game 

Algorithm = set of steps a program takes to complete a task 
Problems in cs have solutions that do the job well for a given taks 
There is an established body of knowledge to solve specific probelms well 
Part of understanding algos is not just knowing that it exists, but knowing when to apply it, which requires understanding the problem 
Algorithmic thinking is looking at a problem, breaking it into tasks and then deciding which algo is best to solve the problem -> better understanding of complexity and efficiency in programming 
Aim is to break an indurmountable problem into distinct components and indentify the right tools to solve each distinct component 

Game: 
num b/w 1-10. Guess the num. When they guess, tell them if it's too high r low. Winner is the one with the fewest guess. This is an example of searching for a value aka search algos

Important parts of algo thinking 
1. Important to establish the probelm set and what values count as inputs - aka establish the boundaries of the problem 
2. With any given prob, there's no one best solution, rather the aim is to figure out which solution works best for the current prob 
3. The specific val being searched for may not matter as much as where that val lies in the given range 
4. the speed at which the prob is solved really matters 
5. An important first step to def the aglo isn't the algo itself, but the prob that I want to solve 
6. Algo uses the same steps to solve the prob every time it's solved 

Guideline for algos which makes for good algo thinking 
1. Must have a clear prob statement, input and output  - specifiy how the input is def and what the output looks like when the algo has done it's job 
2. An algos def must contain a specific set of instructions in a specific order - order of execution needs to be clear 
3. Each step msut not be complex, but explicity clear - it shouldn't be able to be broken down into more substeps - each steo should be distinct 
4. Should produce a result to understand the behaviour 
5. Algo should complete and not have an infinite timeline - complete in a finite amount of time 
6. Executing the steps of an algo for a given input must result in the same output every time 

How to approach a problem 
1. Work through each of the guidelines 
2. Break the prob into any possible N of smaller probes where each prob can be clearly def in terms of input and outputs 
3. Def the prob, input and output before def the algo 

Good algos
1. Correctness -if for every val in the input data, always get an output + the algo should always terminate for every algo - proved by mathetical induction which involves writing a specification and correctness proof
2. Efficiency - solve probs fasters and deliver better experience - measured in time and space complexity. time complexity = how long it takes the algo to run - less time = more efficient. Space complexity - how much memory is taken up. Good algos need to have a balance of time and sapce complexity. These are measured using the Big O notation 
- running time of the algo - how long the algo runs for a given set of vals 

Ways to eval algos 
1. Best case 
2. Avg case - avg running time - not accurate
3. Worst case - Max of the range - indicates that the algowill never perform worse than expected 

Types of search algos 
1. Linear search - start at the beginning and then add 1 num  - aka sequential / simple search 
    - it's an algo because of the specificity of how it's def : we start at the beginning of the given range in the list , if the value != target value, move one by 1 sequentially until target is met 
    - input = series of values. output = target value )can be nothing)
    - for a given value of n in a linear search, it will take n tries
    - doesn't matter if value is not sorted 
2. Binary search - values need to be sorted - start in the middle and then dsicard the half that does not satify the condition, keep decreasing by half and evaluating
    - returns postion in the list as to where the target value is as we already know the target value 
    - input = sorted list of values 
    - output = position in the lsit of target value or sme sort of value to indicate that the target val does not exist in the list 
    - determine the middle psotion of the sorted list
    - comapre the middle element to the target element 
    - if elements match, return the middle psotion and the value 
    - if they don't match, check of the element in the middle position is smaller than the target element, if it is, go back to step 2
    - if it's larger than the target element, go back to step 2 where the new list is from the start of the old list and ends in the middle of the old list, and then go to the middle of this new list 
    - repeat these steps until the target element is found or until a sublist contains only one element - if the single element sublist  does not match the target element , end the algo, indicating that the element does not exist in the list
