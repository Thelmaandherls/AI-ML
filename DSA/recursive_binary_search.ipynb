Defining the RECURSIVE binary search algo as a stand alone func

1. A recurisve func is one that calls itself 
- the last else block is what makes this recurisve as the funciton is callinf on itself to execute that block
- the solution is found by going through the function again - this includes a return statement which is vital 
- instead of normal bs which uses pointers to store the data for the first and last position of the list 
- rbs creates a new sublist and passes it into the function continously until the solution is found and then goes through all the clauses within the function 

2. Structure of a recurisve func 
- Calls itself + has a stopping condition 
- Needs a stopping condition - typically the function starts within this stopping condition: a stopping ocndition is called the base case - this is to cover all paths of logic / outcomes that could happen 
- in rbs, the stopping conditions: the first if statement for an empty list and the second being if the midpoint = the target
- num of times a recurisve func calls itself = recurisve depth 
- functional language = favoured = try to avoid data given to a function - do not like to modify vars within the function 
- langs like python don't like python and has a max recursion depth thf it prefers an iterative solution - langs like swift don't have recursion depths

```python
# Recursive binary search RELIES ON A SORTED LIST or might retun none 
# even if the value is within the list - has to be sorted in asc order
# It breaks the array / list into subarrays / sublists 
# it compares the value at the middle of that sublist 
# until the target value is found 
# Therefore we need a way to keep track of the position 
# of the list we're working with - first and lat to point to 
# the beginning and end of the array
# RETURNS A BOOLEAN - normal bs return the index of the target 

def recursive_binary_search(list_being_searched_through, target_value):
    """
    Returns true if the target is found 
    Otherwise returns False 
    """
    # Start with considering what will happen if the list is empty 
    if len(list_being_searched_through) == 0:
        return False 
    # cal midpoint by div the lenght of the list if it's not empty 
    else:
        midpoint = (len(list_being_searched_through)) // 2 
        # If the value at the midpoint is the same as the target, return True
        if list_being_searched_through[midpoint] == target_value:
            return True 
        # if it's less than the target, call the func again and give it the 
        # protion of the list that we want to focus on 
        # create a new sublist by slicing the current list to be the
        # midpoint + 1 to the end of the list 
        # return the sublist 
        else: 
            if list_being_searched_through[midpoint] < target_value:
                return recursive_binary_search(list_being_searched_through[midpoint+1:], target_value)# O(log n) as keep calling the fun w smaller lists 
            # sublist going all the way tot he midpoint 
            else:
                return recursive_binary_search(list_being_searched_through[:midpoint], target_value)



```

```python
# Cleaned up 

# What makes it an algo: must return a value, 
# must complete execution in a finite amount of time, 
# must return the same output for the same given input


def recursive_binary_search(list, target):
    """
    Returns true if the target is found 
    Otherwise returns False 
    """
    if len(list) == 0:
        return False 
    else:
        midpoint = (len(list)) // 2 
        if list[midpoint] == target:
            return True 
        else: 
            if list[midpoint] < target:
                return recursive_binary_search(list[midpoint+1:], target)
            else:
                return recursive_binary_search(list[:midpoint], target)



```
