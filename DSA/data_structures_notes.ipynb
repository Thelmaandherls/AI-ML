- first item in the ll = head of the list, and last = tail - head and tail nodes = specific as list only maintains ref to the head by sometimes keep ref to tail also - this is v important and most of the ops in a ll need to ebe implemented quite differently to an array 
- tail = opp for head and denotes the end of the list - doesn't point to anythung as all nodes point to the next item in the list 
- nodes = self-referential objs andthe def of a node includes the node itself - forms of ll: 
    - singly linked list - each nodes stores ref to next node in list 
    - doubly linked list - each node stores ref to node before and after in list 
- ways data can be added to a ll:
    - to the head - the most recent node can be the head and the first will be the tail - this ll only keeps ref to the head thf will only add new items to the head of the list - prepend 
    - the most recent node is the tail and the first is the head -requires reference to tail node or would have to start at the head and traverse through the whole list to find the tail- append
    - inserting data into the list is more efficient in a ll as opposed to an arry when inserting the data ino the head or the tail - a true insert is when you can insert the data into any point in the list which is the third way

    - arrays all elements after the specific index are shifted

- removing a node 
    - cocurs in constant time but to get to it, have to traverse the entiety of the ll until reach it which makes it a linear time operation as it's the worst case 
- ll = good for learning pruposes as simple and good to implement for variuous ds - much more efficient ds and many langs e.g. java provide implementation of it 
- inserts in ll are interesting as unlike arrays where you insert an element, all elements needs to be shifted, with a ll, just need to change the references to next on a few nodes - as each nodes points to the next one, by swapping out these references we can insert a node at any point in the list at a contant time
