#### Arrays

A structure for storing and recieving data - a grid in space with each cell contianing one data element 

A 1D array = can only be accessed by one index - elements look like they're written in a list - data is stored in one dimension

A 2D array- data is stored in two dimensions therefore we have rows and columns meaning that to access one element, you need to use 2 indexes, one of the row and another for the column - looks like a table 

3D arrays and higher require additional layers for exmaple a 3d array can be thought of as a cube or a set of tables stacked to form a 3d structure 

Arrays are generalised to an arbitrary number in Numpy using ndarray = number of dimensions in the array 

##### Restrictions to most Numpy arrays 

All elemnts must be the same dtype - int, float, bool, list, set, dict etc 
Total size of array cannot be chnaged once created 
Shape must be rectangular e.g. each row of a 2d array must have the same num of cols 

To note:

For array indexing, the col index is last and the row index is second to last - this is generally applicable for all dimensions of array 

Avoid using math terminology for arrays as array multiplication != matrix multiplication - they are fundamentally different
    - array x requires the arrays to be of the same size -> a matrix of the same size by doing element by element multiplication 
    - matrix x requires num of cols in first matrix to match num of rows in second matrix by doing the summated product of the rows and columns

In general, we say: 0-D array = scalar, 1-D = vector, 2-D = matrix, N-D = tensor where N>2 (diff from PyTorch tensor)

##### Additonal notes

2d array = table - rows and cols 
3d array = another layer 
axes in np: 0 = 1d, 1=2d, 2=3d
shape = num of elements along each axis - row, col 
rank = num of axes / dims - 1d = rank 1, 2d = rank 2

np.empty = creates an array without it being intialised 
    numpy.empty(shape, dtype=float, order='C')
np.absolute(arr) = absolute values of each element in the array 

Numpy Binary Operations apply to the array elementwise and a new array is created. We can use all basic arithmetic operators like +, -, /,  etc. In the case of +=, -=, = operators, the existing array is modified.
