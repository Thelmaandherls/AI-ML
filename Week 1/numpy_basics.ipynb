## Arrays 

Attributes - size, shape, memory sunption and data types 
Indexing - getting and setting value of individual array elements 
Slicing - Getting and setting subarrays within a larger array 
Reshaping - Shanging shape of array 
Joining and splitting - combining multiple arrays into one and splitting one array into many 

#### Attributes 

```python
x1 = np.random.randint(10, size=6) # 1D array 
x2 = np.random.randint(10, size=(3, 4)) # 2D array 
x3 = np.random.randint(10, size=(3, 4, 5)) # 3D array 
```

```python
# Each array has these attributes: 
# ndim = number of dimensions 
# shape = size of each dimension 
# size = total size of the array (a x b x c x ....)

print("x3 ndim", x3.ndim)
print("x3 shape", x3.shape)
print("x3 size", x3.size)
print("dtype:", x3.dtype)
```

```python
# itemsize = size in bytes for each array element 
# nbytes = list total size in bytes of the array : itemsize x size 
print("itemsize:", x3.itemsize, "bytes")
print("nbytes:", x3.nbytes, "bytes")
```

#### Concatenation 

```python
x = np.array([1, 2, 3])
y = np.array([3, 2, 1])
np.concatenate([x, y])
```

```python
z = [99, 99, 99]
print(np.concatenate([x, y, z]))
```

```python
grid = np.array([[1, 2, 3],
                 [4, 5, 6]])
np.concatenate([grid, grid])
```

```python
# Concatenate on the second axis (zero-indexed) - aka concatenate on columns 
np.concatenate([grid, grid], axis=1)
```

```python
# Vertical stack - have to have the same number of columns 
np.vstack([x, grid])
```

```python
# Horizontal stack - have to have the same number of rows 
y = np.array([[99], 
              [99]])
np.hstack([y, grid])
```

```python
# Stack on 3rd axis - each row of both objs become cols
z = np.array([[99, 99, 99], 
              [99, 99, 99]])
np.dstack([z, grid])
```

#### Splitting 

```python
x = [1, 2, 3, 99, 99, 3, 2, 1]
x1, x2, x3 = np.split(x, [3, 5]) # split at index 3 and 5
print(x1, x2, x3)
# N split-points, leads to N + 1 subarrays
```

```python
grid = np.arange(16).reshape(4, 4)
grid
```

```python
upper, lower = np.vsplit(grid, [2])
print(f"Upper: {upper}")
print(f"Lower: {lower}")
```

```python
left, right = np.hsplit(grid, [2])
print(f"Left: {left}")
print(f"Right: {right}")
```

```python
x = np.arange(16.0).reshape(2, 2, 4)
x
```

```python
np.dsplit(x, 2)
```

```python
np.dsplit(x, np.array([3, 6]))
```
